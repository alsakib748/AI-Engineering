```python
# Topic modelling helps us uncover the main themes that run through a collection of documents
# Latent Dirichlet Allocation (LDA)
# 1. Prepare the text
# 2. Build the structures that LDA expects
# 3. Train the model
# Gensim is a python library designed for text analysis
```

```python
import pandas as pd
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
import gensim
import gensim.corpora as corpora
```

```python
data = pd.read_csv("news-articles.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
articles = data['content']
```

```python
# take just the content of the article, lowercase and remove punctuation
articles = articles.str.lower().apply(lambda x: re.sub(r"([^\w\s])","",x))

# stop word removal
en_stopwords = stopwords.words('english')
articles = articles.apply(lambda x: ' '.join([word for word in x.split() if word not in (en_stopwords)]))

# tokenize
articles = articles.apply(lambda x: word_tokenize(x))

# stemming (done for speed as we have a lot of text)
ps = PorterStemmer()
articles = articles.apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
articles
```

```python
dictionary = corpora.Dictionary(articles)
print(dictionary)
```
