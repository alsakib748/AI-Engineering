```python
# N-grams help us analyze the relationship between neighboring words
# N-gram = a sequence of N tokens
# N=1: unigrams ; e.g: 'I','love','NLP' 
# N=2: bigrams ; e.g: 'I love','Love NLP'
# N=3: trigrams ; e.g: 'I love NLP'
```

```python
import nltk
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tokens = [
    'the', 'rise', 'of', 'artificial', 'intelligence','has','led','to','significant','advancements','in',
    'natural','language','processing','computer','vision','and','other','fields','machine','learning','algorithms',
    'are','becoming','more','sophisticated','enabling','computers','to','perform','complex','taks','that','were',
    'learning','neural','networks','have','become','even','more','powerful','capable','of','processing','vast','amounts',
    'of','data','and','learning','from','it','in','ways','that','were','not','possible','industries','from','healthcare',
    'to','finance','to','transportation','and','its','impact','is','only','set','to','grow','in','the','years','to','come'
    ]
print(tokens)
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens,1)).value_counts())
print(unigrams)
```

```python
unigrams[:10].sort_values().plot.barh(color="lightsalmon", width=.9, figsize=(12,8))
plt.title("10 Most requently Occuring Unigrams")
```

```python
bigrams = (pd.Series(nltk.ngrams(tokens,2)).value_counts())
print(bigrams[:10])
```

```python
trigrams = (pd.Series(nltk.ngrams(tokens,3)).value_counts())
print(trigrams[:10])
```
